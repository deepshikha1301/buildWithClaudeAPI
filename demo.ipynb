```python
#load environment variables
from dotenv import load_dotenv
load_dotenv()
```

```python
#create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
#helper function to append messages to the conversation
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)   

#make a request to the API
def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages
    }
    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text
    
    
```

```python
#make starting list for messages
messages = []

#add initial user message
add_user_message(messages, "Define quantum computing in one sentence")
messages

#call the chat function to get a response from the model
response = chat(messages)
response

#add the assistant's response to the messages list
add_assistant_message(messages, response)

#add user's follow-up message
add_user_message(messages, "Write another sentence that expands on the definition")

#call the chat function again to get a response from the model
response = chat(messages)
response
```

```python
#adding a system prompt to the chat function
system_prompt = """You are a helpful assistant. 
                   You will answer the user's questions with clear and concise explanations. 
                   Also, you will give users helpful sources to read more about the topics."""

response =chat(messages, system=system_prompt)
response
```
